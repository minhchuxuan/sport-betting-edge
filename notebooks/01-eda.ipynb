# Exploratory Data Analysis
## Football Match Data (2019-2024)

This notebook explores the cleaned dataset and provides initial insights.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
```

## Load Data


```python
df = pd.read_parquet('../data/processed/matches_clean.parquet')
print(f"Shape: {df.shape}")
print(f"Date range: {df['Date'].min()} to {df['Date'].max()}")
df.head()
```

## Match Outcomes Distribution


```python
result_counts = df['FTR'].value_counts()
print("Match Results:")
print(result_counts)
print(f"\nHome Win Rate: {result_counts['H'] / len(df) * 100:.1f}%")
print(f"Draw Rate: {result_counts['D'] / len(df) * 100:.1f}%")
print(f"Away Win Rate: {result_counts['A'] / len(df) * 100:.1f}%")

# Plot
plt.figure(figsize=(8, 5))
result_counts.plot(kind='bar', color=['#2E86AB', '#A23B72', '#F18F01'])
plt.title('Match Outcome Distribution')
plt.xlabel('Result (H=Home, D=Draw, A=Away)')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()
```

## Goals Analysis


```python
df['TotalGoals'] = df['FTHG'] + df['FTAG']
print(f"Average Home Goals: {df['FTHG'].mean():.2f}")
print(f"Average Away Goals: {df['FTAG'].mean():.2f}")
print(f"Average Total Goals: {df['TotalGoals'].mean():.2f}")

# Plot
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].hist([df['FTHG'], df['FTAG']], bins=range(0, 10), label=['Home', 'Away'], alpha=0.7)
axes[0].set_title('Home vs Away Goals')
axes[0].set_xlabel('Goals')
axes[0].legend()

axes[1].hist(df['TotalGoals'], bins=range(0, 12), alpha=0.7)
axes[1].set_title('Total Goals per Match')
axes[1].set_xlabel('Total Goals')
axes[1].axvline(df['TotalGoals'].mean(), color='red', linestyle='--')
plt.show()
```

## Key Insights

1. **Home Advantage**: Home teams win more frequently
2. **Goals**: Average ~2.7 goals per match
3. **Data Quality**: 100% completeness on core fields
